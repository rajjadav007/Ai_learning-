Calculate BMI

```python
weight=float(input("enter the weight"))
height=float(input("enter the height"))

height_cm=height/100

bmi=weight/(height_cm**2)

print ("yout bmi is",bmi)

if bmi < 18.5:
    print ("you are the underwirght")

elif bmi < 25:
    print ("you are the fit")

elif bmi < 35:
    print ("you are overweight")
```

Calculate percentage of 5 subjects


```python
maths=float(input("enter the maths marks"))
science=float(input("enter the science marks"))
ss=float(input("enter the ss marks"))
hindi=float(input("enter the hindi marks"))
statstics=float(input("enter the statstics marks"))

total=maths+science+ss+hindi+statstics

percentage=(total/500)*100

print ("total marks",total)
print ("percentage of the student is",percentage)

if percentage >80:
    print ("your grade is A")
elif percentage >60:
    print ("your grade is B")
elif percentage >50:
    print ("your grade is C")
elif percentage >40:
    print ("your grade is D")
else :
    print ("you are fail")
```
