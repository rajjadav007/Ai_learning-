```python
import numpy as np 
import pandas as pd
```

## Series 

```python
country=['india','UK','USA','nepal','srilanka']

pd.Series(country)
```

```python
runs=[12,43,56,78,43,100]

pd.Series(runs)
```

```python
marks=[56,89,67,34,90]
subject=['maths','science','hindi','gujarati','java']

pd.Series(marks,index=subject)
```

```python
detail=pd.Series(marks,index=subject,name="Raj ke marks")
detail
```

## Series from dict()

```python
marks={
    'marks':56,
    'science':89,
    'hindi':67,
    'gujarati':34,
    'java':90,
}

full_marks=pd.Series(marks)

full_marks
```
