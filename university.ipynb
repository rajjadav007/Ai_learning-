```python
def factoreal(s):
    if s==0 or s==1:
        return 1
    else:
        return s*factoreal(s-1)
    # print (factoreal(5))
factoreal(12)
```

```python
def prime(n):
    cnt=0
    for i in range(1,n+1):
        if (n%i==0):
            cnt=cnt+1
    if (cnt==2):
        return True
    else:
        return False
    
print(prime(9))
```
