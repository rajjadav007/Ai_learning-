Find quotient and remainder

```python
a=int(input("enter the number of "))
b=int(input("enter the number of 2"))

quotient = a//b
remainder = a%b

print ("the quotient is ",quotient)
print ("the remainder is ",remainder)
```

Calculate simple interest

```python
p=int(input("enter the amount "))
r=float(input("enter the rate of the amount"))
n=float(input("enter the how many years "))

simple_intrest=(p*r*n)/100

print ("your simple intrest is ",simple_intrest)
```

```python
marks=float(input("enter the total marks"))
obtained=float(input("enter the how many mark you get"))

percenteage=(obtained*100)/marks

print ("your percentage is ",percenteage)
```
